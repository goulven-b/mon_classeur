<h2>Qui est Jean-Marc Jancovici ?</h2>

## Qui est Jean-Marc Jancovici ?

jean-Marc Jancovici est un ingénieur français, né le 13 février 1962. diplômé à l'école de polytechnique et de Telecom de Paris, il est spécialisé dans la question du réchauffement climatique et de la dépendance aux énergies fossiles de nos sociétés. il créer le bilan carbone en 2004, et cofonde carbone 4, une entreprise chargée de fournir des bilans carbone pour réduire l'émission de CO2. il crée également le Shift projet, une association qui travaille sur la transition énergétique et la dépendance aux énergies fossiles.

## Quelles sont ses thèses ?

il défend une idée principale, la société d'aujourd'hui et selon lui relié aux énergies fossiles. il faudrait réduire notre consommation énergétique et développer les énergies renouvelables dites basses carbone. malgré qu'il soit pour ce type d'énergie, il est en revanche contre le solaire et l'éolien car cela prendrait trop de place pour le rapport d'énergie produite. c'est pourquoi la solution la plus favorable, selon lui, serait de favoriser le nucléaire, car cela produit une quantité d'énergie maximale dans un espace réduit en émettant, presque aucun gaz à effet de serre. selon lui la transition énergétique est importante car il prévoit un pic pétrolier d'ici 2030, qui conduirait à une grande chute de la consommation.

## Quelle est son actualité ?

Jean-Marc Jancovici prend aujourd'hui souvent la parole sur des médias, tel que RTL par exemple, pour analyser le mouvement étudiant et le décrit comme symptôme de la diminution des flux énergétiques. Il crée également le projet "en commun accord", qui consiste à tirer au sort 150 personnes pour débattre sur la question  de la séparation des énergies fossiles et de proposer des solutions concrètes, en vue des élections présidentielles de 2027.

## fait-il l'unanimité ?

Non, Jean-Marc Jancovici fait débat encore aujourd'hui. il est reconnu pour sa capacité à expliquer les enjeux climatiques, grâce à sa BD le monde sans fin pour le grand public. en revanche, selon certains écologistes il favorisait trop le nucléaire en oubliant parfois les accidents et les déchets radioactifs. Il néglige aussi parfois le potentiel du solaire et de l'éolien. pour lui, la décroissance économique et certaine avec la fin des énergies fossiles, ce qui fait débat car il oublie les nouvelles technologies et le principe de la croissance verte.
