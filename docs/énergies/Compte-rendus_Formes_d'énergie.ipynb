###### Note 2 : La puissance et l'énergie : 

- La puissance caractérise la rapidité ou la vitesse à laquelle l'énergie est transformée d'une forme à une 
autre, ou transférée d'un système à un autre (transfert d’énergie). La puissance d'un transfert énergétique 
est l’énergie fournie ou consommée pendant une seconde. L'énergie s'exprime en Joule (J) et la puissance 
s'exprime en Watt (W).

#### 1. Calculer le courant moyen consommé en vol :

#### 2. Calculer la capacité totale de la charge électrique

###### Par définition, lorsque les cellules sont associées en dérivation, les capacités s'additionnent.
- on sait qu'il y a 3 cellules
- et que chaque cellule a une capacité de $334 \;\mathrm{mAh}$

Ainsi :

$C_{\text{totale}} = 3 \times 334 \;\mathrm{mAh}$

$C_{\text{totale}} = 1002 \;\mathrm{mAh} = 1,002 \;\mathrm{Ah}$

#### 3. En déduire l'autonomie du drone

###### Par définition : $t = \frac{C}{I}$
- on sait que $C = 1,002 \;\mathrm{Ah}$
- et que $I = 4,95 \;\mathrm{A}$

Ainsi :

$t = \frac{1,002 \;\mathrm{Ah}}{4,95 \;\mathrm{A}} \approx 0,202 \;\mathrm{h}$

En minutes :
$0,202 \times 60 \approx 12,1 \;\mathrm{min}$

Pour conclure, l'autonomie de ce drone est donc d'environ $12 \;\mathrm{min}$.
