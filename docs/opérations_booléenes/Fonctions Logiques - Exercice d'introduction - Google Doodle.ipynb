Quelles étaient les raisons de cet hommage rendu par Google ?

> Cet hommage est rendu a Georges Boole qui est né le 8 novembre 1815. Il a crée l'algèbre boléen.

Qu'est-ce qu'une variable booléenne ?

> un type de variable en informatique et en mathématiques qui ne peut prendre que deux valeurs mutuellement exclusives

Qu'elles sont les principales fonctions (opérations) booléennes ?

    > La manipulation de 0 et de 1 avec :
    - ET : renvoie vrai si A et B ont comme valeur 1
    - OR : renvoie vrai si A ou B ont comme valeur 1
    - NON : inverse la valeur d'une condition. si A NON B est est vrai : A = 1 et B = 0
    - XOR : comme OR mais pour les 2 conditions : rencoi vrai si A = 0 et B =1

Expliquer en observant le gif animé du doodle ci-dessus, les conditions pour que :

- le G s'allume en bleu  :

> que x et y vallent 1

- le deuxième o s'allume en jaune :

> si y ou x vallent 1

- le l s'allume en vert :

> si y vaut 0

- le e s'allume en rouge :

> si x vaut 0

- le premier o s'allume en rouge :

> si x vaut 1 et y vaut 0 ou que x vaut 0 et y vaut 1
